Resumen Ejecutivo Ampliado (Visión de Gestión y Arquitectura)
El capítulo «CRISP-DM: el plano maestro» establece que el fracaso de la mayoría de las iniciativas de ciencia de datos no proviene de una mala elección de algoritmos, sino de la falta de un proceso estructurado e ingenieril. CRISP-DM (Cross-Industry Standard Process for Data Mining) se presenta no como un dogma rígido, sino como una guía de referencia universal e independiente de la tecnología que organiza el caos inherente a la experimentación con datos.

El valor central del capítulo radica en abordar el ciclo de vida del proyecto desde el rigor de la ingeniería de software:

Gestión del ciclo de vida iterativo: A diferencia del modelo en cascada (waterfall), CRISP-DM reconoce que el aprendizaje en fases avanzadas altera las decisiones tomadas en etapas tempranas. Las flechas entre fases no son unidireccionales; representan bucles de retroalimentación crítica.

Alineación estratégica (Negocio vs. Técnica): Define un puente claro entre las necesidades comerciales del cliente y las especificaciones técnicas del sistema. Previene el sesgo de construir soluciones técnicamente sofisticadas pero inútiles para la operación real.

Estructura operacional de 6 fases:

Comprensión del negocio: Determinación de objetivos, evaluación de recursos, riesgos, costes y formulación de las preguntas de ciencia de datos asociadas a métricas de impacto (ROI, reducción de churn, etc.).

Comprensión de los datos: Recolección inicial, descripción, exploración y verificación de la calidad del dato. Busca detectar fallos de origen, datos sesgados o faltantes antes de invertir tiempo en desarrollo.

Preparación de los datos: Selección de variables (feature selection), limpieza, transformación, ingeniería de características (feature engineering) y estructuración del dataset final. Suele consumir entre el 70% y 80% del esfuerzo total.

Modelado: Selección de técnicas algorítmicas, diseño del plan de pruebas (validación cruzada, partición train/test) y ajuste de hiperparámetros.

Evaluación: Contraste riguroso del modelo frente a los criterios de negocio iniciales (no solo métricas de error técnico) para decidir si el modelo es apto para producción o si se requiere reevaluar alguna fase previa.

Despliegue: Integración del modelo en la infraestructura de producción, automatización de pipelines, documentación, transferencia de conocimiento y diseño de planes de monitoreo continuo.
