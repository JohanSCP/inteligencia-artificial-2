1. Biología a Redes: Inspiración NaturalOrigen: La ciencia se basa en la estructura biológica de la neurona (cuerpo celular, dendritas y axón) para modelar sistemas computacionales.   Mapeo: Las señales recibidas por las dendritas se convierten en entradas (inputs), el cuerpo celular procesa la información y el axón transmite la respuesta.   2. El Perceptrón: La Unidad FundamentalUnidad Lógica Umbral (TLU): Creada en 1957, combina entradas ($x_1, x_2, x_3$) multiplicadas por pesos ($w_1, w_2, w_3$). Aplica una función lineal $z = w^T x + b$ y posteriormente una función escalón para determinar la salida.   Limitación histórica: Los perceptrones simples únicamente pueden resolver problemas linealmente separables, fallando ante datos o patrones más complejos.   3. Retropropagación y Perceptrón Multicapa (MLP)Arquitectura MLP: Supera las limitaciones del perceptrón simple organizando las neuronas en capa de entrada (Input layer), capas ocultas (Hidden layer) y capa de salida (Output layer).   Proceso de Retropropagación (Backpropagation): Es el método mediante el cual la red aprende:   Adelante (Forward pass): El modelo realiza predicciones con los datos.   Medir Error: Se calcula la pérdida (loss) comparando la predicción con el valor real.   Pase Inverso (Backward pass): Se aplica la regla de la cadena para propagar el error hacia atrás.   Ajustar: Se utiliza el descenso de gradiente para actualizar los pesos y reducir el error.   4. Procesamiento de Datos y Conjuntos de EntrenamientoFashion MNIST: Se utiliza como un conjunto de datos estándar moderno que reemplaza al clásico MNIST para probar modelos de clasificación de imágenes.   Aplanamiento de datos (Flattening): Las matrices o cuadrículas de píxeles bidimensionales se transforman en vectores unidimensionales (secuencias continuas de entradas) antes de pasar por las capas densas de la red neuronal.   5. Ideal para apilar capas de forma consecutiva e intuitiva.   API Funcional:Topología: Compleja.   Flexibilidad: Alta. Permite la creación de redes avanzadas como las estructuras Wide & Deep que conectan múltiples rutas de entrada directamente con capas de salida mediante concatenación (Concat).   API de Subclases (Modelado Dinámico):Topología: Dinámica.   Flexibilidad: Ilimitada. Ofrece control total mediante programación orientada a objetos para implementar comportamientos dinámicos e investigativos personalizados.   6.
