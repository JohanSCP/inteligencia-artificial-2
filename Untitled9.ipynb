1. Reto de los gradientes

Durante el entrenamiento de una red neuronal, los gradientes indican cómo modificar los pesos para reducir el error.

Desvanecimiento del gradiente: los valores del gradiente se vuelven muy pequeños y las primeras capas aprenden muy lentamente.
Explosión del gradiente: los valores se vuelven demasiado grandes y el entrenamiento se vuelve inestable.
Es un problema especialmente importante en redes profundas.
2. Inicialización

Es la forma en que se asignan los valores iniciales de los pesos de la red.

Una buena inicialización permite que el aprendizaje comience de manera estable y ayuda a evitar gradientes demasiado pequeños o grandes.

3. Funciones de activación

Permiten que una red neuronal pueda aprender relaciones no lineales.

Algunas funciones comunes son:

Sigmoid: produce valores entre 0 y 1.
Tanh: produce valores entre -1 y 1.
ReLU: devuelve 0 para valores negativos y el valor original para positivos.
Leaky ReLU: similar a ReLU, pero permite un pequeño valor para entradas negativas.

ReLU y sus variantes son muy utilizadas en redes profundas.

4. Normalización

Busca mantener los valores que pasan por la red en rangos adecuados para facilitar y estabilizar el entrenamiento.

La más conocida es Batch Normalization, que normaliza las activaciones de las capas utilizando estadísticas del lote de datos.

Sus beneficios incluyen:

Entrenamiento más estable.
Posibilidad de utilizar tasas de aprendizaje mayores.
Reducción de ciertos problemas relacionados con los gradientes.
5. Transferencia

La transferencia de aprendizaje (Transfer Learning) consiste en utilizar una red que ya fue entrenada para una tarea y adaptar sus conocimientos a otra tarea relacionada.

Por ejemplo:

Una red entrenada para reconocer diferentes objetos puede adaptarse para identificar tipos específicos de vehículos.

Esto permite ahorrar tiempo, datos y recursos computacionales.

6. Preentrenamiento

Es el proceso de entrenar inicialmente una red utilizando un conjunto de datos grande antes de utilizarla para la tarea específica.

Después se puede realizar fine-tuning (ajuste fino) para adaptar el modelo al problema concreto.

Por ejemplo:

Preentrenamiento → Modelo general → Ajuste fino → Tarea específica
